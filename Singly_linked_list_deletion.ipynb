```python
class Node:
    def __init__(self, data=None):
        self.data = data
        self.next = None


class SLinkedList:
    def __init__(self):
        self.head = None

    def Atbegining(self, data_in):
        NewNode = Node(data_in)
        NewNode.next = self.head
        self.head = NewNode

    def RemoveNode(self, Removekey):
        HeadVal = self.head

        # Case 1: Head node itself holds the key to be removed
        if HeadVal is not None and HeadVal.data == Removekey:
            self.head = HeadVal.next  # Change head
            HeadVal = None  # Free the old head
            return

        # Case 2: Search for the key to be removed in the rest of the list
        prev = None
        while HeadVal is not None and HeadVal.data != Removekey:
            prev = HeadVal
            HeadVal = HeadVal.next

        # If key was not present in linked list
        if HeadVal is None:
            return

        # Unlink the node from linked list
        prev.next = HeadVal.next
        HeadVal = None  # Free the node

    def LListprint(self):
        printval = self.head
        while printval:
            print(printval.data)
            printval = printval.next

llist = SLinkedList()
llist.Atbegining("Mon")
llist.Atbegining("Tue")
llist.Atbegining("Wed")
llist.Atbegining("Thu")
llist.RemoveNode("Tue")
llist.LListprint()
```
