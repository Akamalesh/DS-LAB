```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class CreateList:
    def __init__(self):
        self.head = None
        self.tail = None

    def add(self, data):
        newNode = Node(data)

        if self.head is None:
            self.head = newNode
            self.tail = newNode
            newNode.next = self.head
        else:
            self.tail.next = newNode
            self.tail = newNode
            self.tail.next = self.head

    def deleteEnd(self):
        if self.head is None:
            return
        elif self.head == self.tail: # If there's only one node
            self.head = None
            self.tail = None
        else:
            current = self.head
            while current.next != self.tail:
                current = current.next
            self.tail = current
            self.tail.next = self.head

    def display(self):
        current = self.head
        if self.head is None:
            print("List is empty")
            return
        else:
            print(current.data)
            while current.next != self.head:
                current = current.next
                print(current.data)
        print("\n")


# The class name 'CircularLinkedList' was redundant and the main execution logic
# is placed outside the class for clarity, as it uses an instance of CreateList.

cl = CreateList()
cl.add(1)
cl.add(2)
cl.add(3)
cl.add(4)
print("Original List:")
cl.display()

while cl.head is not None:
    cl.deleteEnd()
    print("Updated List:")
    cl.display()
```
