```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class CircularLinkedList:
    def __init__(self):
        self.last = None

    def addToEmpty(self, data):
        if self.last is not None:
            # If list is not empty, original code returns self.last without modifying.
            return self.last

        temp = Node(data)
        self.last = temp
        self.last.next = self.last
        return self.last

    def addBegin(self, data):
        if self.last is None:
            return self.addToEmpty(data)

        temp = Node(data)
        temp.next = self.last.next
        self.last.next = temp
        return self.last

    def addEnd(self, data):
        if self.last is None:
            return self.addToEmpty(data)

        temp = Node(data)
        temp.next = self.last.next
        self.last.next = temp
        self.last = temp
        return self.last

    def addAfter(self, data, item):
        if self.last is None:
            print(item, "not present in the list (list is empty)")
            return None

        p = self.last.next  # Start from the first node
        found = False
        # Iterate through the circular list
        while True:
            if p.data == item:
                temp = Node(data)
                temp.next = p.next
                p.next = temp
                if p == self.last:  # If we added after the last node, the new node becomes the new last.
                    self.last = temp
                found = True
                break
            p = p.next
            if p == self.last.next:  # If we have traversed a full circle and are back at the start, item not found
                break

        if not found:
            print(item, "not present in the list")
        return self.last  # Return the potentially updated last pointer

    def traverse(self):
        if self.last is None:
            print("List is empty")
            return

        temp = self.last.next
        while True:
            print(temp.data, end=" ")
            temp = temp.next
            if temp == self.last.next:
                break
        print() # Add a newline for better formatting

if __name__ == '__main__':
    llist = CircularLinkedList()

    print("--- Initializing List ---")
    last = llist.addToEmpty(6)
    print(f"List after addToEmpty(6): ", end="")
    llist.traverse()

    last = llist.addBegin(4)
    print(f"List after addBegin(4): ", end="")
    llist.traverse()

    last = llist.addBegin(2)
    print(f"List after addBegin(2): ", end="")
    llist.traverse()

    last = llist.addEnd(8)
    print(f"List after addEnd(8): ", end="")
    llist.traverse()

    last = llist.addEnd(12)
    print(f"List after addEnd(12): ", end="")
    llist.traverse()

    last = llist.addAfter(10, 8) # Add 10 after 8
    print(f"List after addAfter(10, 8): ", end="")
    llist.traverse()

    # Test adding after an item that doesn't exist
    last = llist.addAfter(100, 99)
    print(f"List after addAfter(100, 99): ", end="")
    llist.traverse()
```
