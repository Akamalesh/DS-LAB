```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class CreateList:
    def __init__(self):
        # Initialize with dummy nodes as per original code's apparent intent
        self.head = Node(None)
        self.tail = Node(None)
        self.head.next = self.tail
        self.tail.next = self.head

    def add(self, data):
        newNode = Node(data)
        # If the list effectively contains only the dummy nodes
        if self.head.data is None:
            self.head = newNode
            self.tail = newNode
            self.tail.next = self.head # For a single element circular list
        else:
            self.tail.next = newNode
            self.tail = newNode
            self.tail.next = self.head # Maintain circularity

    def display(self):
        # If the list effectively contains only the dummy nodes
        if self.head.data is None:
            print("List is empty")
            return

        print("Nodes of the circular linked list: ")
        current = self.head
        while True:
            print(current.data)
            current = current.next
            if current == self.head:
                break

cl = CreateList()
cl.add(1)
cl.add(2)
cl.add(3)
cl.add(4)

cl.display()
```
